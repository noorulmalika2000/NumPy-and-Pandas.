```python
#Part 1: NumPy
#Import libraries
import numpy as np
import pandas as pd
```

```python
#Create an 1D array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
#Inspect the array
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
#Array operations
# Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temperatures_f)

# Max, min and mean
print("Maximum:", temperatures_w1.max())
print("Minimum:", temperatures_w1.min())
print("Mean:", temperatures_w1.mean())
```

```python
# First three days
print("First 3 days:", temperatures_w1[:3])

# Weekend (last two days)
print("Weekend:", temperatures_w1[-2:])

# Middle three days
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
#Create a 2D array
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],   # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]    # Week 2
])
print(temperatures)
```

```python
#Inspect and slice the 2D array
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total elements:", temperatures.size)
# Each week
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])
# Weekend of both weeks
print("Weekend of both weeks:")
print(temperatures[:, -2:])
```

```python
#Part 2: Pandas Series
#Create the Series
marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
#Indexing and slicing
# Integer position: 1st rank student
print("1st rank mark:", marks.iloc[0])
# loc: top 3 ranks using labels
print("Top 3 ranks:")
print(marks.loc['Rank1':'Rank3'])
# iloc: 3rd ranked student
print("3rd rank mark:", marks.iloc[2])
# Boolean mask: marks greater than 90
print("Ranks with marks > 90:")
print(marks[marks > 90])
```

```python
#Manipulate the Series
# Change 1st rank mark to 100
marks['Rank1'] = 100
print(marks)
# Remove the last rank
marks = marks.drop('Rank5')
print(marks)
# CGPA = marks / 10
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
#Part 3: Pandas DataFrame
#Create the DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

```python
#Basic information
print("Head:")
print(transactions.head())
print("\nTail:")
print(transactions.tail())
print("\nShape:", transactions.shape)
print("Column names:", list(transactions.columns))
print("\nData types:")
print(transactions.dtypes)
print()
transactions.info()
```

```python
#Only two columns
transactions[['ProductCategory', 'Amount']]
```

```python
#Last 3 columns
transactions.iloc[:, -3:]
```

```python
#Filter rows (North and Amount > 200)
transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
```

```python
#Value counts
print(transactions['ProductCategory'].value_counts())
```

```python
#Unique regions
print(transactions['Region'].unique())
```

```python
#Group by region (mean amount)
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#Modify Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

```python
#Add the Discount column
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```

```python
#Remove the row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
transactions
```

```python
#Delete the Discount column
transactions = transactions.drop(columns='Discount')
transactions
```
